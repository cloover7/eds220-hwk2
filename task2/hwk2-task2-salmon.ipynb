```python
# FINAL CODE
# import data
import pandas as pd
import numpy as np
catch_data = pd.read_csv('data/salmon_data.csv')

# Remove notes column
catch_data = catch_data.drop('notesRegCode', axis = 1)

# reassign 'I' to '1'
catch_data.loc[catch_data['Catch'] == 'I', 'Catch'] = '1'

# reassign type to int
catch_data['Catch'] = catch_data['Catch'].astype('int64')
catch_column_type = catch_data['Catch'].dtype
catch_column_type

# group by region
avg_region = catch_data.groupby('Region')['Catch'].mean().sort_values()
avg_region
```
